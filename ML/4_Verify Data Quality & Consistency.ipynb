# 14 Verify Data Quality & Consistency

## 14.1 Load Dataset

```python
import pandas as pd

DATA_DIR = r"A:\pc\Desktop\Code\data science\sadop\Data"

feature_dataset_path = rf"{DATA_DIR}\ml_features.csv"
df = pd.read_csv(feature_dataset_path)
df.head()
```

```python
# Check for nulls
missing = df.isnull().sum()
print("Missing values per column:\n", missing)
```

## 4.2 cheking for duplicated 

```python
duplicates = df.duplicated().sum()
print("Duplicate rows:", duplicates)
```

## 4.4 Duplicate Removal

```python
df_clean = df.drop_duplicates().reset_index(drop=True)
print("New shape:", df_clean.shape)
```

```python
df_clean.to_csv(r'A:\pc\Desktop\Code\data science\sadop\Dataml_features_clean.csv', index=False)
print(f"✅ ML features dataset verified and saved: ml_features_clean.csv")
```
