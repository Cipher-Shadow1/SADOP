```python
import pandas as pd

path = r"A:\pc\Desktop\Code\data science\sadop\Data\slow_query_metrics.csv"
```

```python

df = pd.read_csv(path)
print(f"Before : {len(df)} rows")

# 1 – drop all rows where timed_out == 1
df = df[df["timed_out"] != 1]
print(f"After  : {len(df)} rows  ({len(df)} kept, timed-out rows removed)")

# 2 – drop the timed_out column entirely
df = df.drop(columns=["timed_out"])

# 3 – save back to the same file
df.to_csv(path, index=False)
print(f"✅ Saved clean dataset → {path}")
```
